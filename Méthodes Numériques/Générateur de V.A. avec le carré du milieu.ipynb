# Générateur de V.A. avec l'algorithme Middle-Square

## 1. Définition
Le Middle-Square est un générateur de nombres pseudo-aléatoires (PRNG) proposé par John von Neumann en 1949. Il repose sur l'extraction des chiffres centraux du carré de la graine précédente.

## 2. Process
Soit $X_n$ un nombre entier de $2k$ chiffres. Pour obtenir $X_{n+1}$ :

1.  **Élévation au carré :** Calculer $Y = X_n^2$. Le résultat aura jusqu'à $4k$ chiffres.
2.  **Padding (Remplissage) :** Si $Y$ a moins de $4k$ chiffres, ajouter des zéros à gauche pour atteindre la taille requise (ex: `67*67 = 4489`, taille 4, ok. Si `10*10 = 100`, voir `0100`).
3.  **Extraction :** Retirer les $k$ chiffres de poids fort (gauche) et les $k$ chiffres de poids faible (droite).
4.  **Résultat :** Les $2k$ chiffres restants forment $X_{n+1}$.

## 3. Implémentation
Pour une graine de $N$ chiffres (base 10) :

* **Opération 1 (Truncature Droite - "La Hache") :**
    `temp = carre // 10^(N/2)`
    *Division entière pour supprimer les chiffres de droite.*

* **Opération 2 (Truncature Gauche - "Le Tamis") :**
    `next_val = temp % 10^N`
    *Modulo pour ne garder que les N derniers chiffres.*

## 4. Exemple de Trace (N=2)
Graine $X_0 = 67$ ($k=1$, on cherche 2 chiffres centraux).

| Étape | Calcul | Valeur | Description |
| :--- | :--- | :--- | :--- |
| **Carré** | $67^2$ | `4489` | Explosion de l'information |
| **Op 1 (// 10)** | $4489 // 10^1$ | `448` | Suppression du chiffre de droite (9) |
| **Op 2 (% 100)** | $448 \pmod{100}$ | `48` | Conservation des 2 derniers chiffres |
| **Sortie** | $X_1$ | `48` | Nouvelle graine |

## 5. Défauts/Problématiques
L'algorithme est instable mathématiquement.
* **Convergence vers Zéro :** Si la séquence atteint un nombre dont le carré est petit (ex: $0 \le X < \sqrt{10^k}$), la séquence s'effondre vers $00...0$ et y reste indéfiniment.
* **Cycles Courts :** La période de répétition est statistiquement très courte par rapport à l'espace des possibles.

**Obsolète :** Remplacé par le LCG (Linear Congruential Generator) pour résoudre le problème de convergence.

```python
#TOUR 1
seed = 67
print(f"Départ 1: {seed}")

print("-" * 20)

print("Etape 1: Mise au carré")
carre = seed * seed   #On passe de 2 à 4 chiffres
print(f"{carre} = {seed} * {seed}") 
print(f"Carré: {carre}")

print("-" * 20)

print("Etape 2: On garde le milieu")
sans_droite = carre // 10  #On coupe à droite avec la Division Entière (// 10)
print(f"Coupe de droite : {sans_droite} (Le chiffre de droite a disparu)")

seed1 = sans_droite % 100  #On coupe à gauche avec le Modulo (% 100)
print(f"Coupe de gauche : {seed1} (Le chiffre de gauche a disparu)")

print("-" * 20)

#TOUR 2
print(f"Départ 2: {seed1}") #On repart avec seed1

print("-" * 20)

print("Etape 1: Mise au carré")
carre2 = seed1 * seed1
print(f"{carre2} = {seed1} * {seed1}") 
print(f"Carré: {carre2}")

print("-" * 20)

print("Etape 2: On garde le milieu")
sans_droite2 = carre2 // 10  #On coupe à droite avec la Division Entière (// 10)
print(f"Coupe de droite : {sans_droite} (Le chiffre de droite a disparu)")

seed1 = sans_droite2 % 100  #On coupe à gauche avec le Modulo (% 100)
print(f"Coupe de gauche : {seed1} (Le chiffre de gauche a disparu)")

print("-" * 20)

print(f"RESULTAT : {seed1}")
```
