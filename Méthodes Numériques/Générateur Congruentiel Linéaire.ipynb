

## I. Introduction
Méthode la plus ancienne et la plus simple pour générer des nombres pseudo-aléatoires.
Proposé par Derrick Henry Lehmer en 1951, mathématicien américain pionnier de la théorie des nombres et de l'informatique.
Les Nombres "Magiques": Les chiffres spécifiques $a = 16807$ et $m = 2^{31}-1$ proviennent d'une publication célèbre de Park & Miller en 1988 (intitulée "Random Number Generators: Good Ones Are Hard To Find"). Ils ont prouvé que ce couple de chiffres précis offrait le meilleur mélange possible pour les ordinateurs 32 bits de l'époque.

## 2. Le Fondement Mathématique

Le principe repose sur l'arithmétique modulaire (l'arithmétique de l'horloge). L'équation fondamentale est :$$X_{n+1} = (a \cdot X_n + c) \pmod m$$

Dans la version Park-Miller, $c = 0$, donc la formule se simplifie en :$$X_{n+1} = (16807 \cdot X_n) \pmod{2147483647}$$

## 3. Comment ça marche (Le concept de "Chaos Déterministe")
L'ordinateur est incapable de hasard. Il suit des règles strictes. L'idée de Lehmer est de créer une suite de nombres qui a l'air chaotique, mais qui est calculée rigoureusement.L'Expansion ($a \cdot X$) : On prend le nombre actuel et on le multiplie par $16807$. Cela projette le point très loin sur la ligne des nombres, créant un écart énorme entre deux points proches.Le Repliement (Modulo $m$) : C'est la clé. Le modulo "coupe" ce grand nombre en tranches de la taille de $m$ et ne garde que le reste. Cela ramène brutalement le nombre au début de l'échelle.

```python
# 1. Paramètres (Constantes de Park-Miller)
m = 2147483647
a = 16807
x = 12345  # La Graine (Seed) : Change ce nombre pour changer le tirage

# 2. La Formule (Générateur Congruentiel Linéaire)
x = (a * x) % m

# 3. Le Résultat (Normalisation entre 0 et 1)
u = x / m

print(u)
```
