# Générateur de V.A. avec l'heure système

**Objectif** : Générer un nombre entier aléatoire entre 1 et 100 sans utiliser de bibliothèque.

→ On utilise uniquement le temps actuel de l'ordinateur (qui change tout le temps) pour créer du "hasard".

```python
from time import time

current_time = time()
print("1. Temps actuel:", current_time)

integer_part = int(current_time)
print("2. Partie entière:", integer_part)

fractional_part = current_time - integer_part
print("3. Partie décimale:", fractional_part)

multiplied = fractional_part * 100
print("4. Multiplié par 100:", multiplied)

integer_after_int = int(multiplied)
print("5. Après int():", integer_after_int)

final_number = integer_after_int + 1
print("6. Nombre aléatoire final (1-100):", final_number)
```
